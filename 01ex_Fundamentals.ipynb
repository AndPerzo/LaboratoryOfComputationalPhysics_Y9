1\. Write the following as a list comprehension

````python
# 1
ans = []
for i in range(3):
    for j in range(4):
        ans.append((i, j))
print (ans)

# 2
ans = map(lambda x: x*x, filter(lambda x: x%2 == 0, range(5)))
print (list(ans))
````

````python
#1 soluzione
"with list comprehension"
ans = [(i**3,j) for i in range(3) if i!=1 for j in range(4)]
print(ans)

#1.1 soluzione
"with list comprehension"
ans = [(i**3,j) for i in range(3) for j in range(4) if i!=1]
print(ans)

#2 soluzione
ans = [x*x for x in range(5) if x%2 == 0]
print(ans)
````

2\. Convert the following function into a pure function with no global variables or side effects

````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````

````python
#2 soluzione

" Mutabili: list, dict (dizionari), set (insiemi), bytearray."
" Immutabili: int, float, bool, str (stringhe), tuple (tuple), frozenset, bytes."
        
def f(alist):
    new_list = alist.copy()
    
    for i in range(5):
        new_list.append(i)
    
    return new_list

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
````
